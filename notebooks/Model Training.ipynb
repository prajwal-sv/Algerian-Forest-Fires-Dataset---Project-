```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python


df=pd.read_csv('../data/Algerian_forest_fires_cleaned_dataset.csv')
```

```python
df.head()
```

```python
df.columns
```

```python
##drop month,day and yyear - axis=1 → tells Pandas to drop columns (instead of rows).
## - inplace=True → applies the change directly to df (no new DataFrame is returned).

df.drop(['day','month','year'],axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Classes'].value_counts()
```

```python
## Encoding 
# this line converts text labels into numeric form:
#- "not fire" → 0
#- anything else → 1




df['Classes']=np.where(df['Classes'].str.contains("not fire"),0,1)
```

```python
df.tail()
```

```python
df['Classes'].value_counts()
```

```python
## Independent And dependent features
# X = df.drop('FWI', axis=1)
# → Creates a new DataFrame 'X' with all columns except 'FWI'.
#   (axis=1 means drop a column, not a row)
#   This becomes your input features for ML.

# y = df['FWI']
# → Extracts the 'FWI' column from df.
#   This becomes your target/output variable.
X=df.drop('FWI',axis=1)
y=df['FWI']
```

```python
X.head()
```

```python
y
```

```python
#Train Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
X_train.shape,X_test.shape
```

```python
## Feature Selection based on correlaltion
X_train.corr()
```

```python
## Check for multicollinearity
plt.figure(figsize=(12,10))
corr=X_train.corr()
sns.heatmap(corr,annot=True)
```

```python
def correlation(dataset, threshold):
    # Initialize an empty set to store names of correlated columns
    col_corr = set()

    # Compute the correlation matrix of the dataset
    corr_matrix = dataset.corr()

    # Loop through each column index
    for i in range(len(corr_matrix.columns)):
        # Compare with all previous columns (to avoid duplicate checks)
        for j in range(i):
            # If absolute correlation value is greater than the threshold
            if abs(corr_matrix.iloc[i, j]) > threshold:
                # Get the column name that is highly correlated
                colname = corr_matrix.columns[i]
                # Add it to the set of correlated columns
                col_corr.add(colname)

    # Return the set of correlated column names
    return col_corr
```

```python
## threshold--Domain expertise
corr_features=correlation(X_train,0.85)
```

```python
## drop features when correlation is more than 0.85 
X_train.drop(corr_features,axis=1,inplace=True)
X_test.drop(corr_features,axis=1,inplace=True)
X_train.shape,X_test.shape
```

## Feature Scaling Or Standardization

```python
from sklearn.preprocessing import StandardScaler
# Import StandardScaler to normalize/standardize features

scaler = StandardScaler()
# Create a scaler object that will standardize data 
# (mean = 0, standard deviation = 1)

X_train_scaled = scaler.fit_transform(X_train)
# Fit the scaler on training data (learn mean & std) 
# and transform it → produces scaled training features

X_test_scaled = scaler.transform(X_test)
# Use the same scaler (with training mean & std) 
# to transform test data → ensures consistency
```

```python
X_train_scaled
```

## Box Plots To understand Effect Of Standard Scaler

```python
plt.subplots(figsize=(15, 5))
plt.subplot(1, 2, 1)
sns.boxplot(data=X_train)
plt.title('X_train Before Scaling')
plt.subplot(1, 2, 2)
sns.boxplot(data=X_train_scaled)
plt.title('X_train After Scaling')
```

## Linear Regression Model

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
linreg=LinearRegression()
linreg.fit(X_train_scaled,y_train)
y_pred=linreg.predict(X_test_scaled)
mae=mean_absolute_error(y_test,y_pred)
score=r2_score(y_test,y_pred)
print("Mean absolute error", mae)
print("R2 Score", score)
plt.scatter(y_test,y_pred)
```

## Lasso Regression

```python
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
lasso=Lasso()
lasso.fit(X_train_scaled,y_train)
y_pred=lasso.predict(X_test_scaled)
mae=mean_absolute_error(y_test,y_pred)
score=r2_score(y_test,y_pred)
print("Mean absolute error", mae)
print("R2 Score", score)
plt.scatter(y_test,y_pred)
```
